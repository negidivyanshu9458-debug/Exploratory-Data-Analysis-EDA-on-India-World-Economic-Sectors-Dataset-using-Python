# HISTOGRAM 

# GDP DISTRIBUTION 

```python
#step 55: # Display GDP distribution using histogram
# It shows how GDP values are spread across the dataset
# Histogram shows the distribution of numerical data by dividing values into ranges (bins).
plt.figure(figsize=(8,5))
plt.hist(df["GDP_Value_USD_Bn"], bins=8,edgecolor="black")
plt.title("GDP Distribution")
plt.xlabel("GDP Value(USB Billion)")
plt.ylabel("Frequency")
plt.show()
```

# HISTOGRAM WITH KDE

```python
#step 56:# Add KDE curve to understand GDP distribution pattern
sns.histplot(
df["GDP_Value_USD_Bn"],
kde=True,edgecolor="black"
)
```

# box plot 

```python
# step 57: A box plot is a visualization used to show the distribution of data and detect outliers.
# It displays the minimum, maximum, median, and quartile values of a dataset.
# Create box plot to detect GDP outliers
# Helps identify extremely high or low values
plt.figure(figsize=(6,4))
sns.boxplot(data=df,y="GDP_Value_USD_Bn")
plt.title("GDP Value Distribution and Outlier Detection")
plt.ylabel("GDP Value (USD Billion)")
plt.show()
```

# scatter plot

# GDP vs Employment

```python
# step 58:A Scatter Plot is a graph used to show the relationship between two numerical variables. 
# It represents data points using dots.
# Hue represents different economic sectors
plt.figure(figsize=(10,6))
sns.scatterplot(data=df,x="Employment_%",y="GDP_Value_USD_Bn",hue="Sector ",s=100,alpha=1)
plt.title("GDP vs Employment",fontsize=19)
plt.grid(True)
plt.legend(title="Economic sector ")
plt.show()
```

# LINE PLOT

# YEAR VS GDP

```python
#step 59: Line Plot is a data visualization technique used to show trends or changes over time by connecting data points with a line
# Create line graph showing GDP trend over years
# Calculate average GDP by year
year = df.groupby("Year")["GDP_Value_USD_Bn"].mean()
plt.figure(figsize=(10,5))
plt.plot(year.index,year.values,marker="o")
plt.title("Year vs Average GDP Trend")
plt.xlabel("Year")
plt.ylabel("Average GDP (USD Billion)")
plt.grid(alpha=0.3)
plt.show()
```

# count plot

# countries

# HEATMAP

# CORRELATION

```python
#step 61: Select only numerical columns from dataset
# Calculate correlation between numerical variables
corr = df.select_dtypes(include=np.number).corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".4f",linewidths=0.6,square=True)
plt.title("Correlation Matrix of Economic Dataset", fontsize=14)
plt.show()
```

# VIOLIN PLOT

```python
#  step 63: Create a professional violin plot to compare GDP distribution across different sectors
# The violin plot represents the distribution and variation of GDP values across different economic sectors. 
# It helps identify sector-wise GDP concentration, spread, median values, and possible outliers."
plt.figure(figsize=(12,6))
sns.violinplot(data=df,x="Sector ",y="GDP_Value_USD_Bn",inner="box",palette="Set2")
plt.title("GDP Distribution Across Different Economic Sectors")
plt.xlabel("Economic Sector")
plt.ylabel("GDP Value (USD Billion)")
plt.grid(alpha=0.3)
plt.show()
```
